```python

# Name: Mohd Shafique
# Roll No: 24BAD074
# PART A: Loading and Configuring a Pre-trained LLM

!pip install -q transformers torch accelerate

from transformers import AutoTokenizer, AutoModelForCausalLM
import torch

model_name = "Qwen/Qwen2.5-0.5B-Instruct"

# Load tokenizer and model
tokenizer = AutoTokenizer.from_pretrained(model_name)

model = AutoModelForCausalLM.from_pretrained(
    model_name,
    torch_dtype="auto",
    device_map="auto"
)

# Prepare sample prompt
prompt = "Explain Artificial Intelligence in simple words."

messages = [
    {"role": "user", "content": prompt}
]

input_text = tokenizer.apply_chat_template(
    messages,
    tokenize=False,
    add_generation_prompt=True
)

inputs = tokenizer(input_text, return_tensors="pt").to(model.device)

# Generate response
with torch.no_grad():
    outputs = model.generate(
        **inputs,
        max_new_tokens=100,
        temperature=0.7,
        do_sample=True,
        top_p=0.9
    )

# Display response
response = tokenizer.decode(
    outputs[0][inputs["input_ids"].shape[1]:],
    skip_special_tokens=True
)

print("User:", prompt)
print("Chatbot:", response)
```

```python

# Roll No: 24BAD074
# PART B: Chatbot Functionality

def generate_response(user_query, history=None):

    if history is None:
        history = []

    messages = [
        {
            "role": "system",
            "content": "You are a helpful AI assistant. Give clear and simple answers."
        }
    ]

    # Add previous conversation
    for item in history:
        messages.append(item)

    # Add current user query
    messages.append({
        "role": "user",
        "content": user_query
    })

    # Convert messages into model input
    text = tokenizer.apply_chat_template(
        messages,
        tokenize=False,
        add_generation_prompt=True
    )

    inputs = tokenizer(
        text,
        return_tensors="pt"
    ).to(model.device)

    # Generate chatbot response
    with torch.no_grad():
        outputs = model.generate(
            **inputs,
            max_new_tokens=150,
            do_sample=True,
            temperature=0.7,
            top_p=0.9
        )

    response = tokenizer.decode(
        outputs[0][inputs["input_ids"].shape[1]:],
        skip_special_tokens=True
    )

    return response


# Test chatbot
query = "What is Machine Learning?"
answer = generate_response(query)

print("User:", query)
print("Chatbot:", answer)

query2 = "Give me one example."
answer2 = generate_response(
    query2,
    [
        {"role": "user", "content": query},
        {"role": "assistant", "content": answer}
    ]
)

print("\nUser:", query2)
print("Chatbot:", answer2)
```

```python

# Roll No: 24BAD074
# PART C: Gradio Chatbot Interface

import gradio as gr

def chatbot(message, history):

    history = history or []

    response = generate_response(
        message,
        history
    )

    return response


# Create Gradio interface
demo = gr.ChatInterface(
    fn=chatbot,
    title="AI Chatbot",
    description="Ask any question to the Transformer-based LLM."
)

# Launch application
demo.launch(share=True)
```
