## Part A: Dataset preparation and text preprocessing
Load the Brown Corpus, select the news category, lowercase and tokenize the text, remove punctuation and stop words, and lemmatize the remaining words.

```python
# Mohd Shafique R B | Roll No: 24BAD074
import re
import nltk
from nltk.corpus import brown, stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer

for resource in ["brown", "punkt", "punkt_tab", "stopwords", "wordnet", "omw-1.4"]:
    nltk.download(resource, quiet=True)

category = "news"
raw_sentences = brown.sents(categories=category)
original_text = " ".join(" ".join(sentence) for sentence in raw_sentences[:8])
stop_words = set(stopwords.words("english"))
lemmatizer = WordNetLemmatizer()

tokens = word_tokenize(original_text.lower())
preprocessed_tokens = [
    lemmatizer.lemmatize(token)
    for token in tokens
    if re.fullmatch(r"[a-z]+", token) and token not in stop_words
]

print("ROLL NO: 24BAD074")
print("Original text:\n", original_text[:1000])
print("\nPreprocessed tokens:\n", preprocessed_tokens[:150])
```

## Part B: Vocabulary generation and word-frequency analysis

```python
# Mohd Shafique R B | Roll No: 24BAD074
from nltk import FreqDist
from nltk.corpus import brown

sentences = []
for sentence in brown.sents(categories="news"):
    cleaned = [
        lemmatizer.lemmatize(word.lower())
        for word in sentence
        if re.fullmatch(r"[a-zA-Z]+", word) and word.lower() not in stop_words
    ]
    cleaned = [word for word in cleaned if len(word) > 0]
    if cleaned:
        sentences.append(cleaned)

all_tokens = [word for sentence in sentences for word in sentence]
frequency = FreqDist(all_tokens)
vocabulary = sorted(set(all_tokens))

print("ROLL NO: 24BAD074")
print("Vocabulary size:", len(vocabulary))
print("\nTop 10 most frequent words:")
for word, count in frequency.most_common(10):
    print(f"{word}: {count}")
print("\nFirst 20 frequency entries:", list(frequency.items())[:20])
```

## Part C: Word embedding generation using Word2Vec
Train Skip-gram Word2Vec with 100-dimensional vectors and a context window of 5.

```python
# Mohd Shafique R B | Roll No: 24BAD074
%pip -q install gensim

from gensim.models import Word2Vec

word2vec_model = Word2Vec(
    sentences=sentences,
    vector_size=100,
    window=5,
    min_count=2,
    workers=4,
    sg=1,
    epochs=10,
    seed=42
)

candidate_words = ["government", "people", "year", "state", "market"]
selected_word = next((word for word in candidate_words if word in word2vec_model.wv), None)

print("ROLL NO: 24BAD074")
print("Vocabulary size in Word2Vec:", len(word2vec_model.wv))
if selected_word:
    vector = word2vec_model.wv[selected_word]
    print("Selected word:", selected_word)
    print("Embedding vector:", vector)
    print("Vector dimensions:", vector.shape)
    print("Five most similar words:", word2vec_model.wv.most_similar(selected_word, topn=5))
else:
    print("No candidate query word was found in the trained vocabulary.")
```

## Part D: Word similarity analysis
The code checks that both words in each pair exist in the model before calculating cosine similarity.

```python
# Mohd Shafique R B | Roll No: 24BAD074
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np

candidate_pairs = [
    ("government", "state"),
    ("people", "man"),
    ("year", "day")
]

print("ROLL NO: 24BAD074")
similarity_results = []
for word_a, word_b in candidate_pairs:
    if word_a in word2vec_model.wv and word_b in word2vec_model.wv:
        score = float(cosine_similarity(
            word2vec_model.wv[word_a].reshape(1, -1),
            word2vec_model.wv[word_b].reshape(1, -1)
        )[0, 0])
        similarity_results.append((word_a, word_b, score))
        print(f"Cosine similarity ({word_a}, {word_b}) = {score:.4f}")
    else:
        print(f"Skipped ({word_a}, {word_b}): one or both words are not in vocabulary.")

if similarity_results:
    print("\nMost similar pair:", max(similarity_results, key=lambda item: item[2]))
```
